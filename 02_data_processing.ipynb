```python
#packages
#we import functions using this:
import numpy #numpy is a common package for handling numerical data
```

```python
#reading in data with numpy
numpy.loadtxt(fname='molerat_activity_v1.csv',
              delimiter=",")
#array of variable, "shape" gives information about simensions of array
```

```python
data=numpy.loadtxt(fname='molerat_activity_v1.csv',delimiter=",")
```

```python
print(data.dtype)
#ans shape to confirm we have all data
print(data.shape)
```

```python
#slicing using []
data[0,1] #show the data in the first row and second column
print('the value in the first row, second columnis',data[0,1])
```

```python
#whole rows
data[0,] #first row
print(data[0,])
#whole columns
data[:,0] #first column (all rows is specified by a : with nothing on either side)
print(data[:,0])
```

```python
#slicing ranges (subsets of some rows/columns)
print(data[0:10, ]) #looks at first 10 rows, all columns
print(data[ : ,0:10]) #first columns, all rows (all rows need : to get range)

#IMPORTANT NOTES: the first value is INCLUSIVE , the second value is EXCLUSIVE
#0:10 will print from row/column 0 to row/column 9
```

```python
#these data are naked mole-rat activity
#rows are individuals
#columns are observation sessions
#cells are the count of how many times an individual was active in each session

#what is the mean number across the whole dataset
print(numpy.mean(data))
```

```python
#what about individual mole rats?
numpy.mean(data[0,])#average activity for the first individual
print('activity from individual 1:', numpy.mean(data[0,]))
```

```python
# what about sessions?
numpy.mean(data[:,0]) #mean for the observation session 1
```
